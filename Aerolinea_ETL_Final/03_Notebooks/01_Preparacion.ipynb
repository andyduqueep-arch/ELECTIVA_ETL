### Pipeline de datos de una aerolínea (Apartados 1 a 4)

Arquitectura Medallion sobre **Databricks + Spark + Delta Lake + Unity Catalog**.


#### Paso 0: Configuración de Unity Catalog y carga de los CSV



```python
# Configuración general del proyecto
CATALOG = "aerolinea_trabajo_final"
SCHEMAS = ["landing", "bronze", "silver", "analytics", "gold", "ml"]
RAW_PATH = f"/Volumes/{CATALOG}/landing/raw_files"

spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
spark.sql(f"USE CATALOG {CATALOG}")
for schema in SCHEMAS:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.landing.raw_files")

print(f"Catálogo listo: {CATALOG}")
print(f"Sube los CSV a: {RAW_PATH}")
```

```python
# Verificación de que los CSV están en el volumen
archivos = [f.name for f in dbutils.fs.ls(RAW_PATH)]
print(archivos)
for esperado in ["flights.csv", "airlines.csv", "airports.csv"]:
    assert esperado in archivos, f"Falta subir {esperado} a {RAW_PATH}"
print("Los tres CSV están disponibles.")
```
